## Import libraries

```python
import os

from pandas.api.types import is_string_dtype, is_numeric_dtype

import pandas as pd
import geopandas as gpd

# Display all columns when looking at data frames 
pd.set_option('display.max_columns', None)
```

## Import power plants data

```python
URL = 'https://raw.githubusercontent.com/carmengg/eds-220-book/refs/heads/main/data/power_plants_epsg4269.csv'
power_plants = pd.read_csv(URL)

power_plants.head(3)
```

## Import US states shapefile

```python
file = os.path.join('data','tl_2026_us_state1', 'tl_2026_us_state.shp')
states = gpd.read_file(file)

states.head(3)
```

## Create geodataframe from CVS

Known from the metadata: CRS of the powerplant data has the code EPSG:4269

```python
# Create points from latittude and lonitidue 
points = gpd.points_from_xy(power_plants['longitude'],power_plants['latitude'])

# Create geodataframe 
# Create geodataframe
power_plants = gpd.GeoDataFrame(power_plants,    # Data
                                geometry=points, # Specify geometry column
                                crs='EPSG:4269'  # Specify CRS
                                )
```

## f-strings

```python
print('CRS', power_plants.crs)
```

```python
# Explore crs 
print(f'ellipoid: {power_plants.crs.ellipsoid}')
```

```python
# Any expression can go inside the curly brackets 
print(f"The data has {len(power_plants):,} power_plants")
print(f"Average installed capacity {power_plants['install_mw'].mean():.2f} MW")
```

## `assert`

We want to update the data type of the columns from strings to integers. And we want to check it works without printing things

```python
# Update code columns into integers
for column in code_cols:
    states[column] = states[column].astype('int')
    assert is_numeric_dtype(states[column])
```
